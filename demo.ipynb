```python
from agents.cv_parser_agent import CVParserAgent
from agents.rag_agent import RAGAgent
from agents.cv_generator_agent import CVGeneratorAgent
from agents.cv_feedback_agent import CVFeedbackAgent
from agents.email_formulator_agent import EmailFormulatorAgent
import json
```

```python
cv_parser_agent = CVParserAgent()
rag_agent = RAGAgent()
cv_generator_agent = CVGeneratorAgent()
cv_feedback_agent = CVFeedbackAgent()
email_formulator_agent = EmailFormulatorAgent()
```

```python
user_input = "Hello! I am interested in graphs and everything related to them. For example distributed computing on graphs and GNNs."
pdf_path = "YardenCV.pdf"
```

```python
# parse pdf
response_cv_parser = cv_parser_agent.generate_response_with_pdf(pdf_path)
print(response_cv_parser)
```

```python
# get RAG response
response_rag = rag_agent.generate_response(user_input, response_cv_parser)
print(response_rag)
```

```python
# save responses
with open("response_cv_parser.json", "w") as f:
    json.dump(response_cv_parser, f)
with open("response_rag.json", "w") as f:
    json.dump(response_rag, f)
```

```python
# load responses
with open("response_cv_parser.json", "r") as f:
    response_cv_parser = json.load(f)
with open("response_rag.json", "r") as f:
    response_rag = json.load(f)
```

```python
count = 0
feedback = ""
while count < 2:
    # generate CV
    response_cv_generator = cv_generator_agent.generate_response(user_input, response_cv_parser, response_rag, feedback=feedback)
    print(response_cv_generator)

    # get feedback
    response_cv_feedback = cv_feedback_agent.generate_response(user_input, response_cv_parser, response_cv_generator)
    print(response_cv_feedback)

    # load response_cv_feedback into json
    json_response_cv_feedback = json.loads(response_cv_feedback)
    decision = json_response_cv_feedback["decision"]
    feedback = json_response_cv_feedback["feedback"]
    if decision == "accept":
        break

    print("Rejected CV, generating new one")
    count += 1
```

```python
# formulate email
response_email_formulator = email_formulator_agent.generate_response(user_input, response_cv_generator, response_rag)
print(response_email_formulator)
```
