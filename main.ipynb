```python
from agents.agent import Agent
from agents.cv_parser_agent import CVParser
from agents.rag_agent import RAGAgent
from agents.cv_generator_agent import CVGenerator
from agents.cv_parser_agent import CVParser
from agents.cv_feedback_agent import CVFeedback
from agents.email_formulator_agent import EmailFormulator
import json
```

```python
user_input = "Hi, i'm Rom and I am looking for a superviser!! In the subject of LLMs and Security!!"
pdf_path = "draft_cv.pdf"
```

```python
cv_parser_agent = CVParser()
rag_agent = RAGAgent()
cv_generator_agent = CVGenerator()
cv_feedback_agent = CVFeedback()
email_formulator_agent = EmailFormulator()
```

```python
parsed_cv = cv_parser_agent.generate_response_with_pdf(user_input=user_input, pdf_path=pdf_path)
print(parsed_cv)
```

```python
rag_response = rag_agent.generate_response(user_input, parsed_cv)
print(rag_response)
```

```python
cv_generator_response = cv_generator_agent.generate_response(user_input, parsed_cv, rag_response)
print(cv_generator_response)
```

```python
# save responses to files
# parsed_cv
with open("parsed_cv.txt", "w") as f:
    f.write(parsed_cv)

# rag_response
with open("rag_response.txt", "w") as f:
    f.write(rag_response)

# cv_generator_response
with open("cv_generator_response.txt", "w") as f:
    f.write(cv_generator_response)
```

```python
# load responses from files
with open("parsed_cv.txt", "r") as f:
    parsed_cv = f.read()

with open("rag_response.txt", "r") as f:
    rag_response = f.read()

with open("cv_generator_response.txt", "r") as f:
    cv_generator_response = f.read()
```

```python
accepted = False
feedback = ""
counter = 0
while not accepted and counter < 3:
    counter += 1
    cv_generator_response = cv_generator_agent.generate_response(user_input, parsed_cv, rag_response, feedback)
    print(f"Generated CV: {cv_generator_response}")
    cv_feedback_response = cv_feedback_agent.generate_response(user_input, parsed_cv, cv_generator_response)

    cv_feedback = json.loads(cv_feedback_response)
    decision = cv_feedback["decision"]
    feedback = cv_feedback["feedback"]
    print(f"Desicion: {decision}")
    print(f"Feedback: {feedback}")
    if decision == "accept":
        accepted = True

if not accepted:
    print("CV generation failed, please try again.")
```

```python
email_formulator_response = email_formulator_agent.generate_response(user_input, cv_generator_response, rag_response)
print(email_formulator_response)
```

```python
# Final output: reason + email + cv
print(f"Supervisor reason: {rag_response}")
print(f"Formatted email: {email_formulator_response}")
print(f"Generated CV: {cv_generator_response}")
```
